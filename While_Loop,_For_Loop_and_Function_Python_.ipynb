**3. Implement the Guessing Logic:**

Use a while loop to allow the user to keep guessing until they get the correct
number or run out of attempts.

**Provide feedback to the user for each guess:**

If the guess is out of the valid range (1 to 10), inform the user.
If the guess is greater than the secret number.
If the guess is lower than the secret number.
If the guess is correct, congratulate the user and end the game.

```python
import random

# From Step 1 & 2
secret_number = random.randint(1, 10)
attempts = 0
max_attempts = 3  # you can change to 5 if you want more chances

# Step 3 - Guessing Logic
while attempts < max_attempts:
    guess = int(input(f"Attempt {attempts+1}/{max_attempts} - Enter your guess (1 to 10): "))

    # Condition 1: Out of range check
    if guess < 1 or guess > 10:
        print("Invalid input! Please enter a number between 1 and 10 only.")
        continue  # don't count this as an attempt

    attempts += 1

    # Condition 2: Greater than secret number
    if guess > secret_number:
        print("Too high! Try again.")

    # Condition 3: Lower than secret number
    elif guess < secret_number:
        print("Too low! Try again.")

    # Condition 4: Correct guess
    else:
        print(f"Congratulations! You guessed it in {attempts} attempts.")
        break

# If loop finishes without break (ran out of attempts)
else:
    print(f"Game Over! You ran out of attempts. The secret number was {secret_number}.")
```

**4. Control Statements:**

★Use continue to skip the rest of the loop after informing the user if the guess is out of range.

★Use break to exit the loop when the guess is correct.

★ Use else with the while loop to provide a message like "Better luck next time!"
if the user runs out of attempts without guessing the correct number.

```python
import random

# 1. Set Up the Game
secret_number = random.randint(1, 10)
max_attempts = 5
attempts = 0

print("Welcome to the Number Guessing Game!")
print("You have 5 attempts to guess the number between 1 and 10.\n")

# 2. Game Loop with Control Statements
while attempts < max_attempts:
    guess = int(input("Guess the number (between 1 and 10): "))
    attempts += 1

    # Use continue to skip if out of range
    if guess < 1 or guess > 10:
        print("Your guess is out of range. Please guess a number between 1 and 10.\n")
        continue

    # Check guess
    if guess < secret_number:
        print("Too low. Try again.\n")
    elif guess > secret_number:
        print("Too high. Try again.\n")
    else:
        print(f"Congratulations! You guessed the correct number in {attempts} attempts.")
        break  # Use break to exit when correct

else:
    # This else executes if while loop finishes without break
    print(f"Better luck next time! The correct number was {secret_number}.")
```

**Task 3: Function- BMI Calculator**

Problem Statement: Create a Python program that calculates the Body Mass Index (BMI).
Hint: BMI = weight (kg) / [height (m)]2

**Instructions:**
1. Define a function calculate_bmi(weight, height) that returns the BMI.
2. Prompt the user for their weight (in kg) and height (in meters).
3. Use the function to calculate and display the BMI

```python
# 1. Define a function to calculate BMI
def calculate_bmi(weight, height):
    """Calculates BMI = weight / (height^2)"""
    bmi = weight / (height ** 2)
    return bmi

# 2. Prompt user for input
weight = float(input("Enter your weight in kg: "))
height = float(input("Enter your height in meters: "))

# 3. Use the function to calculate and display BMI
bmi_value = calculate_bmi(weight, height)

print(f"Your BMI is: {bmi_value:.2f}")

# Optional: Category Display
if bmi_value < 18.5:
    print("Category: Underweight")
elif 18.5 <= bmi_value < 24.9:
    print("Category: Normal weight")
elif 25 <= bmi_value < 29.9:
    print("Category: Overweight")
else:
    print("Category: Obese")
```
